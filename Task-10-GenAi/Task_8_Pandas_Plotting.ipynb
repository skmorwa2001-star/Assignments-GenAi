## Task 8: Pandas Plotting (Simple Graphs)

```python
import pandas as pd
```

```python
# Create dataframe

students={
    'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78, 85, 90, 66, 72],
    'Subject':['Math','Math','Science','Science','Math']
}

df=pd.DataFrame(students)
```

```python
df
```

```python
# 1. Plot a bar graph of student names Vs Marks

df.plot(x='Name', y='Marks',kind='bar',title='Student Name Vs Marks')
```

```python
# 2. Plot a line graph of marks

df['Marks'].plot(kind='line',title='Marks Line Graph')

```

```python
# 3. Plot a histrogram of marks

df['Marks'].plot(kind='hist',title='Histogram of Marks')
```
