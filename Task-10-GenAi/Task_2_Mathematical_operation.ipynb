## Task 2: Mathematical Operations on Series

```python
import pandas as pd
```

```python
# Marks
marks=[78,85,90,66,72]

## Series of marks
marks_series=pd.Series(data=marks)
```

```python
# Print Series
marks_series
```

```python
# 1. Add 5 grace marks
print("Add 5 grace marks")
print(marks_series+5)
```

```python
# 2. Subtract 2 marks from all values
print("Subtract 2 marks")
print(marks_series-2)
```

```python
# 3. Multiply all marks by 1.05
print("Multupy all marks by 1.05")
print(marks_series*1.05)
```

```python
# 4. Divide all marks by 2
print("Divide all marks by 2")
print(marks_series/2)
```
