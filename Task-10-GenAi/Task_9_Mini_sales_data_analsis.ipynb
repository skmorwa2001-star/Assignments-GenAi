## Task 9: Mini Use Case: Sales Data Analysis

```python
import pandas as pd
```

```python
# Create a dataframe

sales = {
    'Day':['Mon','Tue','Wed','Thu','Fri'],
    'Revenue':[1200, 1500, 900, 2000, 1800]
}

df_sales=pd.DataFrame(sales)
```

```python
# Print dataframe
df_sales
```

```python
# Perform
 
# 1. Total Revenue

total_revenue=df_sales['Revenue'].sum()

print("Total Revenue:", total_revenue)
```

```python
# 2. Average daily revenue

avg_revenue=df_sales['Revenue'].mean()

print("Average daily revenue:", avg_revenue)
```

```python
# 3. Day with highest revenue

high_revenue=df_sales['Revenue'].max()

print("Day with highest revenue:",high_revenue)
```

```python
# 4. Days where revenue > average

print("Days where revenue > average revenue:\n",df_sales[df_sales['Revenue']>avg_revenue])
```

```python
# 5. Plot revenue vs day

df_sales.plot(x='Day',y='Revenue',kind='line', title='Revenue Vs Day')
```
