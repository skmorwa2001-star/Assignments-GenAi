## Assignment 6: Filtering & Conditional Selection

```python
import pandas as pd
```

```python
# Create dataframe

students={
    'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78, 85, 90, 66, 72],
    'Subject':['Math','Math','Science','Science','Math']
}

df=pd.DataFrame(students)
```

```python
df
```

```python
df.info()
```

```python
## Perform the following filters:

# 1. Students who scored more than 75 marks

print("Students scored more than 75 marks:\n", df[df['Marks']>70])
```

```python
# 2. Students belonging to subject Math

print("Students Beloning to Math:\n", df[df['Subject']=='Math'])
```

```python
# 3. Students who scored more than average marks

print("Students scored more than average marks:", df['Marks'].mean())
```

```python
# 4. Students who failed (marks<70)

print("Students failed:\n",df[df['Marks']<70])
```
