## Task 3: Python Functionalities on Series

```python
import pandas as pd
```

```python
marks=pd.Series([78,85,90,66,72])
```

```python
marks
```

```python
# 1. Find

# Maximum marks
print("Maximum Marks:",marks.max())

# Minimum marks
print("Minimum Marks:",marks.min())

# Sum
print("Sum of Marks:",marks.sum())

# Mean
print("Average Mark:",marks.mean())
```

```python
# 2. Apply a lambda function to check wheather each student has passed

passed=marks.apply(lambda x:x>=70)

print("Student Passed")
print(passed)
```

```python
# 3. Count how many student passed

print("Number of students passed:",passed.sum())
```
