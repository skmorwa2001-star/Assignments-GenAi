## Task 7: Grouping & Basic Analysis

```python
import pandas as pd
```

```python
# Create dataframe

students={
    'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78, 85, 90, 66, 72],
    'Subject':['Math','Math','Science','Science','Math']
}

df=pd.DataFrame(students)
```

```python
df
```

```python
# 1. Find the average marks per subject

print("Average marks per subject:\n", df.groupby('Subject')['Marks'].mean())
```

```python
# 2. Count number of students per subject

print("Count number of students per subject:\n", df.groupby('Subject')['Name'].count())
```

```python
# 3. Find maximum marks per subject

print("Maximum marks per subject:\n", df.groupby('Subject')['Marks'].max())
```
