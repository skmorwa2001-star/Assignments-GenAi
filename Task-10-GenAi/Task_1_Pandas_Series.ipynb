## Task 1: Pandas Series Basics

```python
import pandas as pd
```

```python
# Create series
marks=[78, 85, 90, 66, 72]

## Indexing 
marks_series=pd.Series(data=marks, index=('Maths','English','Chemistry','Physics','Hindi'), name='Marks of Each Subjects')

## Print
print(marks_series)
```

```python
# Access:
## First element
print("First Element:",marks[0])

## Last two elements
print("Last two elements:",marks[-2:])
```
