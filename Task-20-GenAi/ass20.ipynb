### Text 2: Text Preprocessing for Recommendation

```python
from nltk.corpus import stopwords
import re
import pandas as pd
```

```python
stopwords=stopwords.words('english')
```

```python
# Convert into lowercase

data['clean_text']=data['description'].str.lower()
data['Product Name']=data['Product Name'].str.lower()
```

```python
# remove punctuation and special characters
def punc_removal(text):
    if pd.isna(text):
        return ""
    return re.sub(r"[^\w\s]","",text)
```

```python
data['clean_text']=data['clean_text'].apply(punc_removal)
data['Product Name']=data['Product Name'].apply(punc_removal)
```

```python
## remove stopwords

def stopword_remove(text):
    stopword_clean=[]
    for i in text.split():
        if i in stopwords:
            stopword_clean.append("")
        else:
            stopword_clean.append(i)
    return  " ".join(stopword_clean)              
```

```python
data['clean_text']=data['clean_text'].apply(stopword_remove)
data['Product Name']=data['Product Name'].apply(stopword_remove)
```

```python
# Handle missing values

data['clean_text']=data['clean_text'].fillna('').astype(str)
data['Product Name']=data['Product Name'].fillna('').astype(str)
```

## PART 2 - Text Vectorization

### Task 3: Vectorization using TF-Idf

```python
# use tfidfvectorizer() to convert text into vectors

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# set parameters
tf=TfidfVectorizer(max_features=5000,ngram_range=(1,2))
```

```python
matrix_tf=tf.fit_transform(data['clean_text'])
```

```python
print("Shape of TF-IDF",matrix_tf.shape)
```

### Task 4: Similarity Computation

```python
# Compute cosine similarity between all items
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Store similarity matrix
similarity_matrix=cosine_similarity(matrix_tf)
```

```python
print("Shape of similar matrix: ",similarity_matrix.shape)
print(similarity_matrix)
```

```python
'''
3. Explain brief why cosine similarity is used
- Cosine similarity measures the angle between the vectors
- it is useful in recommendation because it primarily used to give pattern/ direction comparision ratheer than length of documents 
'''
```

## PART 4 - Recommendation Logic

### Task 5: Build Recommendation Function

```python
def recommend(product_name, top_n=5):
    product_index=data[data['Product Name']==product_name].index[0]

    similarity_scores=similarity_matrix[product_index]

    product_list=sorted(enumerate(similarity_scores),reverse=True,key=lambda x:x[1])[1:6]

    recommendations=[]

    for index, score in product_list:
        if index!=product_index:
            recommendations.append({'Product Name':data.iloc[index]['Product Name']})

        if len(recommendations) == top_n:
            return recommendations    
```
