### Task 2: Basic Text Cleaning

#### Apply the following basic cleaning steps:

```python
import re
```

```python
# 1. Convert text to lowercase

data['clean_text_basic']=data['reviews'].str.lower()
```

```python
# 2. Remove punctuation

def remove_punctuation(text):
    if pd.isna(text):
        return ""
    return re.sub(r"[^\w\s]","",text)
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_punctuation)
```

```python
# 3. Remove numbers

def remove_number(text):
    if pd.isna(text):
        return 0
    return re.sub(r'\d+',"",str(text))
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_number)
```

```python
# 4. remove extra whitespaces

def remove_extra_whitespace(text):
    if pd.isna(text):
        return ""
    return re.sub(r'\s+',' ',str(text)).strip()
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_extra_whitespace)
```
