### Task 2: Basic Text Cleaning

#### Apply the following basic cleaning steps:

```python
import re
```

```python
# 1. Convert text to lowercase

data['clean_text_basic']=data['reviews'].str.lower()
```

```python
# 2. Remove punctuation

def remove_punctuation(text):
    if pd.isna(text):
        return ""
    return re.sub(r"[^\w\s]","",text)
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_punctuation)
```

```python
# 3. Remove numbers

def remove_number(text):
    if pd.isna(text):
        return 0
    return re.sub(r'\d+',"",str(text))
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_number)
```

```python
# 4. remove extra whitespaces

def remove_extra_whitespace(text):
    if pd.isna(text):
        return ""
    return re.sub(r'\s+',' ',str(text)).strip()
```

```python
data['clean_text_basic']=data['clean_text_basic'].apply(remove_extra_whitespace)
```

### Task 3: Removing Noise

```python
# 1. Remove URLs

def remove_url(text):
    if pd.isna(text):
           return ""
    return re.sub(r'https?://\S+|www\.\S+',"",text)
```

```python
data['clean_text_advanced']=data['clean_text_basic'].apply(remove_url)
```

```python
data['clean_text_advanced'][0]
```

```python
# 2. Remove emails addresses

def remove_email(text):
    email_pattern=r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
    return re.sub(email_pattern,'',text)
```

```python
text="Contact me sunil@gmail.com for more information"
```

```python
remove_email(text)
```

```python
# 3. Remove HTML tags

def remove_html(text):
    return re.sub(r'<.*?>',"",text)
```

```python
text="<h1> Hello my name is sunil</h>"
```

```python
remove_html(text)
```

```python
# 4. Remove special characters & emojis

def remove_emojis(text):
    emoji_pattern=re.compile(
        "["
        u"\U0001F600-\U0001F64F" # EMOTICONS
        u"\U0001F300-\U0001F5FF" # SYMBOLS AND PICTOGRAPHS 
        u"\U0001F680-\U0001F6FF" # TRANSPORT AND MAP
        u"\U0001F1E0-\U0001F1FF" # FLASH 
        "]+",
        flags=re.UNICODE 
    )
    return emoji_pattern.sub(r'',text)
```

```python
text="Happy birthday 🎂"
```

```python
remove_emojis(text)
```

### Task 4: Handling Stopwords

```python
# 1. Load stopwords using NLTK

from nltk.corpus import stopwords
```

```python
english=stopwords.words('english')
```

```python
def without_stopwords(text):
    without_stopword=[]
    for i in text.split():
        if i in english:
            without_stopword.append("")
        else:
            without_stopword.append(i)
        
    return " ".join(without_stopword)
```

```python
# 2. Remove stopwords from clean_text_advanced

data['text_no_stopwords']=data['clean_text_advanced'].apply(without_stopwords)
```

### Task 5: Handling Repeated characters & slang (Optional)


```python
slang_dict={"u":"you","gr8":"great","btw":"by the way","pls":"please"}
```

```python
def normalize_slang(text):
    # Normailize repeated characters
    text=re.sub(r'(.)\1{2,}',r'\1',text)
    # Replace slang words
    words= text.split()
    words=[slang_dict.get(word.lower(),word) for word in words]

    return " ".join(words)
```

```python
text="Sooo good! U are gr8 btw"
```

```python
normalize_slang(text)
```
