# Task 1: Basic function: Price After Discount

```python
def apply_discount(price, discount_percent=5):
    if discount_percent>60:
        discount_percent=60

    final_price = price-(price*discount_percent/100)
    return final_price
```

```python
print(apply_discount(1000, 70))
```

```python
print(apply_discount(500))
```

# Task 2: Recursive function

```python
def factorial(num):
    if num==0 or num==1:
        return 1
    elif num>1:
        return num*factorial(num-1)
    else:
        print("Error: Factotial is not defined for negative number")
```

```python
print(factorial(5))
```

```python
print(factorial(0))
```

```python
factorial(-3)
```

### Create another lambda to compute final price after GST + discount together

```python
# for GST(%)
gst_percent=18
gst = lambda price:price + (price*gst_percent/100)

# Discount(%)
discount_percent=10
discount = lambda price:price-(price*discount_percent/100)
 
price=1000
final_price = discount(gst(price))

print("the price is:",price)
print("the final price is :",final_price)

```

# Task 4 - Using map() : Apply GST to list of prices

```python
# GST lambda function
gst_percent=18
gst = lambda prices:prices + (prices * gst_percent/100)

#Given
prices=[100, 250, 400, 1200, 50]

# Apply GST after price
prices_with_gst=list(map(gst, prices))

print("The Original prices:",prices)
print("Prices after GST:",prices_with_gst)
```

# Task - 5 Using filter(): Filter Expensive products

```python
# Given 
prices = [100, 250, 400, 1200, 50, 2000, 850]

# prices greater 
prices_greater_500=list(filter(lambda price:price>500, prices))

# prices less than equal to 500
prices_equal_less_500=list(filter(lambda price:price<=500, prices))
```

```python
print("The prices greater than 500:",prices_greater_500)
print("The prices less than or equal to 500:",prices_equal_less_500)
```

# Task 6 - Combined Utility Function

```python
def process_prices(prices):
    
    # Discount(10%)
    discount_prices=list(map(lambda price:price-(price*0.10), prices))

    # discount prices only apply above 300
    filtered_prices=list(filter(lambda price:price>300, prices))

    return discount_prices, filtered_prices

# Test
prices = ([100, 500, 900, 50, 750])


print("Apply discount:",discount_prices)
print("Only discounted prices above on 300:",filtered_prices)
```

# Task 7- Mini Problem: Menu using functions

```python
#list
prices_list=[]

#Add prices to list
def add_price(prices_list, price):
    prices_list.append(price)


# Averge price
def get_average_price(prices_list):
    if len(prices_list)==0:
        return 0

    return sum(prices_list)/len(prices_list)

# Maximium value price
def get_max_price(prices_list):
    if max(prices_list)==0:
        return 0

    return max(prices_list)

# Menu
while True:
    print("\n1. Add Value")
    print("2. Show average value")
    print("3. Show highest price")
    print("q. Quit")     

    chose=input("Enter your number")

    if chose=="1":
        price=float(input("Enter price:", ))
        add_price(prices_list,price)

    elif chose=="2":
        print("Average Value:",get_average_price(prices_list))

    elif chose=="3":
        print("Highest Value:",get_max_price(prices_list))

    elif chose=="q":
        print("Quit")
        break

    else:
        print("Invalid input")                         
```
