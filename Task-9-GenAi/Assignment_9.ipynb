# Task 3: Important Numpy Mathematical Formulas

```python
# Given
values=np.array([2,4,6,8,10])

# Calculate:

## Square root
square_value=np.sqrt(values)

## Exponential
expo_value=np.exp(values)

## logarithm
log_value=np.log(values)

## Sum  of elements
sum_value=np.sum(values)

## Cumulative sum of elements
cum_value=np.cumsum(values)



# Print:

print("Square root of each element:",square_value,"\n")
print("Exponential of each element:",expo_value,"\n")
print("Natural logarithmm of each element:",log_value,"\n")
print("Sum of all Elements:",sum_value,"\n")
print("Cumulative sum of element:",cum_value)
```

## Task 4: Aggregation Operations

```python
# Given
data = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

# Find:

## Row wise sum
row_wise=np.sum(data,axis=1)

## Column wise sum
col_wise=np.sum(data,axis=0)

## Minimium value
min_value=np.min(data)

## Maximium value
max_value=np.max(data)

## mean
mean_value=np.mean(data)
```

```python
# Print
print("Row wise Sum:",row_wise)
print("Column wise Sum:",col_wise)
print("Minimum Value:",min_value)
print("Maximum Value:",max_value)
print("Mean Value:",mean_value)
```

# Task 5: Statistical Operations (Core Focus)

```python
# Given 
marks=np.array([78,85,90,66,72,88,95,60])

# Calculate

## Mean Value
mean_mark = np.mean(marks)

## Median Value
median_mark = np.median(marks)

## Variance
var_mark = np.var(marks)

## Standard Deviation
std_mark = np.std(marks)

## Minimum & Maximum
min_mark = np.min(marks)
max_mark = np.max(marks)

## Range between min - max marks
range_mark = max_mark - min_mark
```

```python
# Print
print("Mean Mark:",mean_mark)
print("Median Mark:",median_mark)
print("Variance:",var_mark)
print("Standard Deviation:",std_mark)
print(f"Minimum Mark: {min_mark}","&",f"Maximum Mark: {max_mark}")
print("Range Marks:",range_mark)
```

# Task 6: Percentiles & Sorting

```python
## Given
marks=np.array([78,85,90,66,72,88,95,60])

# 1. Sort the array
sorted_marks=np.sort(marks)
print("Sorted Marks:",sorted_marks)


# 2. find:
## 25th percentile
print("25th Percentile:",np.percentile(marks,25))
## 50th percentile
print("50th Percentile:",np.percentile(marks,50))
## 75th percentile
print("75th Percentile:",np.percentile(marks,75))


# 3. Count the students scored above the above marks
average=np.mean(marks)
count=np.sum(marks>average)

print("Average Marks:",average)
print("Students Above Average:",count)
```

# Task 7: Mini Use Case : Sales Analysis

```python
# Given daily sales data
sales=np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

# Perform:
## Total weekly sales
total_sales=np.sum(sales)

## Average daily sales
average_sales=np.mean(sales)

## Highest & Lowest sales day
high_sales=np.max(sales)
low_sales=np.min(sales)

## Standard Devitaion
std_sales=np.std(sales)

## Identify days where sales were above average
above_avg=np.where(sales>average_sales)[0]+1  # +1 because days starts from 1
```

```python
# Print Perform:
print("Total weekly sales:",total_sales)
print("Average daily Sales:",average_sales)
print(f"Highest sale day: {high_sales}","&",f"Lowest sale day: {low_sales}")
print("Standard deviation of sales:",std_sales)
print("Days With Sales above Average:",above_avg)
```
