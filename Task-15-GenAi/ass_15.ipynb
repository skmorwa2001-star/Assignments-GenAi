```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## PART 1 - Regression Alogorithm

### Task 1 : Linear Regression

```python
# Read the dataset from JSON file

data=pd.read_json('product_data copy.json')
```

```python
data.head()
```

```python
# 1. Select numercial target variable

Feature_Data=data[['markdown_percentage','current_price','Total Value','Purchase Year','Purchase Day','Brand Length of text','Word Count']]
Target_Data=data[['original_price']]
```

```python
X=Feature_Data
Y=Target_Data
```

```python
# Check any missing value present in dataset
X.isnull().sum()
```

```python
Y.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
# 2. Split the data into training and testing sets

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
# import LinearRegression model

from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
# 3. Train a Linear Regression model using scikit-learn
model.fit(X_train,Y_train)
```

```python
# 4. Make Prediction on test data
y_predict=model.predict(X_test)
```

```python
y_predict[:5]
```

```python
# 5. Plot Actual vs Predicted values

plt.figure(figsize=(7,4))
plt.title("Plot Actual vs Predicted Values")
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.scatter(Y_test,y_predict)
```
