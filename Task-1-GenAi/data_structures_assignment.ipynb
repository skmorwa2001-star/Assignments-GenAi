# Task 3 : Product Pricing

```python
# Create a dictionary with product names and prices
price_dict={
    "Laptop":70000,
    "Books":500,
    "Monitor":15000,
    "Keyboard":2000,
    "Mouse":500,
    "Pen":50
}
```

```python
print("Original Dictionary:",price_dict)
```

```python
# Add a new product
price_dict["Tablet"]=30000
print("After adding new product:", price_dict)
```

```python
# Update price of an existing product
price_dict["Mouse"]=700
print("After updating Mouse price:", price_dict)
```

```python
# Remove a product
del price_dict["Pen"]
print("After removing Pen:", price_dict)
```

```python
# Calculate average price
total=0
for price in price_dict.values():
    total=total + price
average_price=total/len(price_dict)
print("Average price of products:", average_price)    
```

```python
# Optical: find max and min price
max_product=max(price_dict, key=price_dict.get)
min_product=min(price_dict, key=price_dict.get)

print("Product with maximum price:", max_product)
print("Product with minimum price:", min_product)
```

# Task 4: Combined operations

```python
# Products list
products=["Laptop","Books","Monitor","Keyboard","Mouse","Pen",]
```

```python
# Price dictionary
price_dict={
    "Laptop":70000,
    "Books":500,
    "Monitor":15000,
    "Keyboard":2000,
    "Mouse":500,
    "Pen":50
}
```

```python
# Category list
categories=["Electronics","Education","Electronics","Electronics","Electronics","Stationery",]
```

```python
# Create a catalog 
catalog = []

for i in range(len(products)):
    catalog.append((products[i], price_dict[products[i]], categories[i]))
    print("Catalog:", catalog)
    
```

```python
# Create category_to_products dictionary
category_to_products = {}

for item in catalog:
    product = item[0]
    category = item[2]
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(product)

    print("Category to products mapping:", category_to_products)
```

```python
# find maxi and mini products
max_product=max(price_dict, key=price_dict.get)
min_product=min(price_dict, key=price_dict.get)
print("Product with maximum price:", max_product)
print("Product with minimum price:", min_product)
```
