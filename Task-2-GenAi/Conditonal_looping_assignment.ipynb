## Task 1 : Discount rules

```python

# Input User
order_amount=int(input("Enter amount of order : "))

# Apply discount
if order_amount>=2000:
    discount_percent = 15

elif order_amount>=1500 and order_amount<2000:
    discount_percent = 10

elif order_amount>=1000 and order_amount<1500:
    discount_percent = 7

#Calculate discount
discount=(order_amount*discount_percent)/100
subtotal=order_amount-discount 

#Optical : Add 5% tax
tax=subtotal*0.05
final_amount=subtotal+tax


print("---Bill Summary---")
print("Original_amount:",order_amount)
print("Discount(%) : ",discount_percent)
print("Discount amount :",discount)
print("Subtotal :",subtotal)  
print("Tax(5%) :",tax)
print("Final Amount :",final_amount)       
```

## Task 2: Process Multiple Orders (for loop)

```python
# lists of orders
orders = [1200, 2500, 800, 1750, 3000]

total_revenue=0

print("orders Amount\tDiscount%\tFinal Amount")

for order in orders:

    if order >= 2000:
        discount_percent = 15

    elif order >= 1500 and order<2000:
        discount_percent = 10

    elif order >= 1000 and order<1500:
        discount_percent = 7
    else:
        discount_percent = 0    

    #Calculate discount
    discount = (order * discount_percent) / 100
    final_value=order-discount 

    # Total Revenue
    total_revenue += final_value
    
    print(f"{order},\t\t{discount_percent}%\t\t{final_value}")

print("Total Revenue After discounts:", total_revenue)
```
