## PART 1 - One - Hot Encoding (Text Level)

### Task 1: Manual one-hot encoding

```python
import pandas as pd
```

```python
data=pd.read_csv('final_cleaned_data.csv')
```

```python
# 1. Select 5 sentences 

sentences=data['final_clean_text'].dropna().iloc[:5].tolist()

for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

```python
# 2. Build a vacabulary of unique words

vocabulary=sorted(set(
    word 
    for sentence in sentences 
    for word in sentence.split()))
```

```python
# 3. Create one hot vectors

one_hot_vectors=[]

for sentence in sentences:
    words = set(sentence.split())

    vector=[1 if word in words else 0 for word in vocabulary]

    one_hot_vectors.append(vector)
```

```python
# Display

one_hot_df=pd.DataFrame(one_hot_vectors, columns = vocabulary)
```

```python
print(vocabulary)
```

```python
one_hot_df
```

### Task 2: One-Hot Encoding using Scikit-learn

```python
# 1. Use countvectorizer(binary=True)

from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(binary=True)
```

```python
# 2. Fit and transform the selected sentences

encoded_matrix=cv.fit_transform(sentences).toarray()
```

```python
# 3. Print

print("Vocabulary :",cv.vocabulary_)

print("Encoded matrix :",encoded_matrix)
```

## PART 2 - Bag of Words (BoW)

### Task 3: Bag of Words Representation

```python
# Use the CountVectorizer

from sklearn.feature_extraction.text import CountVectorizer
```

```python
bow=CountVectorizer()
```

```python
# 2. Fit the entire dataset

texts=data['final_clean_text'].dropna().astype(str)
```

```python
print("Number of texts: ",len(texts))
```

```python
# 3. Transform text into BoW vectors
#  
bow_matrix=bow.fit_transform(texts)
bow_vocabulary=bow.get_feature_names_out()
```

```python
# Display:

print("Vocabulary Size:",len(bow_vocabulary))
print("Sample feature vectors:",bow_matrix[:5].toarray())
```

### Task 4: Understanding Word Frequency

```python
# Get Vocabulary

words=bow.get_feature_names_out()
```

```python
# Calculate total frequency of each word

word_frequency=bow_matrix.sum(axis=0)
```

```python
# Convert into 1D array

word_frequency=word_frequency.A1
```

```python
# Create DataFrame

frequency_df=pd.DataFrame({'word':words, 'frequency':word_frequency})
```

```python
#  Sort by frequency

frequency_df=frequency_df.sort_values(by='frequency',ascending=False)
```

```python
frequency_df
```

```python
# Top 10 most frequent words

top_10=frequency_df.head(10)
```

```python
print("Top 10 most frequent words:\n", top_10)
```

```python
# Least frequent words
least_frequent=frequency_df[frequency_df['frequency']==frequency_df['frequency'].min()]
```

```python
print("Least frequent words:\n",least_frequent)
```

Explain how BoW captures frequency information?
- The Bag of Words (BoW) model captures frequency information by converting text into numerical vectors where each dimension represents a known word, and the value in that dimension represents how many times that word appears. It throws away all grammar , context ,and word order treating the document like a literal bag of disconnected words

## PART 3 - N-Grams

### Task 5 : Unigrams, Bigrams & Trigrams

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# 1. Create BoW using: 

unigram=CountVectorizer(ngram_range=(1,1))
bigram=CountVectorizer(ngram_range=(2,2))
trigram=CountVectorizer(ngram_range=(3,3))
```

```python
unigram_matrix=unigram.fit_transform(texts)
unigram_vocabulary=unigram.get_feature_names_out()
```

```python
bigram_matrix=bigram.fit_transform(texts)
bigram_vocabulary=bigram.get_feature_names_out()
```

```python
trigram_matrix=trigram.fit_transform(texts)
trigram_vocabulary=trigram.get_feature_names_out()
```

```python
# 2. Compare:

print(f"Vocabulary Size :", len(unigram_vocabulary),len(bigram_vocabulary),len(trigram_vocabulary))

print("Sample feature vectors :\n", unigram_matrix[:5].toarray(),bigram_matrix[:5].toarray(),trigram_matrix[:5].toarray())
```
