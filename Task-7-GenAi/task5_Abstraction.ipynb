# Task 5 : Abstraction (Using Abstract Base Class)

```python
from abc import ABC, abstractmethod

# Abstract class

class Payment(ABC):

    # Abstract method
    @abstractmethod
    def process_payment(self,amount):
        pass


# Abstract subclass 1
class CreditCardPayment(Payment):

    def process_payment(self,amount):
        print(f"Credit Payment of",amount, "proccessed successfully")


# Abstract subclass 2       
class UPIPayment(Payment):

    def process_payment(self,amount):
        print("UPI Payment of",amount, "proccessed successfully")

```

```python
# Create objects of subclass

pay1=CreditCardPayment()
pay2=UPIPayment()
```

```python
# Test 1
pay1.process_payment(500)
```

```python
# Test 2
pay2.process_payment(2000)
```
