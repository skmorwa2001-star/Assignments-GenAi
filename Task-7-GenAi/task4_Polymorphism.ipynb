## Task 4 : Polymorphism

```python
# Parent Class
class Product:
    def __init__(self, name, price, category):
        self.name=name
        self.price=price
        self.category=category

    def get_info(self):
        print("Products Details")
        print("Name:", self.name)
        print("Price:",self.price)
        print("Category:",self.category)

```

```python
## First Child class
class Laptop(Product):
    def __init__(self,name,price,category,ram):
        super().__init__(name,price,category)
        self.ram=ram

    # Override the get.info()
    def get_info(self):
        print("\n-----Laptop Details------")
        print("Name :", self.name)
        print("Price :",self.price)
        print("Category :", self.category)
        print("Ram :", self.ram, "GB")    
```

```python
## Second Child Class
class Mobile(Product):
    def __init__(self,name,price,category,camera):
        super().__init__(name,price,category)
        self.camera=camera

    # Overriding 
    def get_info(self):
        print("\n-----Mobile Details------")
        print("Name:", self.name)
        print("Price :", self.price)
        print("Category :", self.category)
        print("Camera :",self.camera,"MP")    
```

```python

## Create Objects
laptop=Laptop("Asus Laptop",60000,"Electronic",32)
mobile=Mobile("Motorola",25000,"Electrronic",64)
```

```python
# Create a list
products=[laptop, mobile]
```

```python
# Loop through objects and call get_info()
for item in products:
    item.get_info()
```
