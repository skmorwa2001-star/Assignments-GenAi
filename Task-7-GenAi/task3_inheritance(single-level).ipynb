# Task 3 : Inheritance (Single-Level)

```python
## Parent class
class Product:
    def __init__(self, name, price, category):
        self.name=name
        self.price=price
        self.category=category

    def get_info(self):
        print("Products Details")
        print("Name:", self.name)
        print("Price:",self.price)
        print("Category:",self.category)
```

```python
## Child class inheriting from Product
class ElectronicProduct(Product):
    ## Add warrenty to child class
    def __init__(self,name,price,category,warrenty_years):
        super().__init__(name,price,category)
        self.warrenty_years=warrenty_years


    # Override get_info() method
    def get_info(self):
        print("Products Details")
        print("Name:", self.name)
        print("Price:",self.price)
        print("Category:",self.category)
        print("Warreenty Years :", self.warrenty_years)
            
```

```python
## Create a child class
p1=ElectronicProduct("Smart Phone",20000,"Electronic",4)
```

```python
## Call overridden method
p1.get_info()
```
