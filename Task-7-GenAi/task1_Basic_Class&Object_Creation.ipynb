## Task 1: Basic Class & Object Creation

```python
# Create class
class Product:
    def __init__(self, name, price, category):
        self.name=name
        self.price=price
        self.category=category

    # Print the product details
    def get_info(self):
        print("Products Details")
        print("Name:", self.name)
        print("Price:",self.price)
        print("Category:",self.category)
```

```python
# Create objects
p1=Product("Milk",50,"food")
p2=Product("Laptop",80000,"Electronics")

# Print the product details
print(p1.get_info())
print()
print(p2.get_info())
```
