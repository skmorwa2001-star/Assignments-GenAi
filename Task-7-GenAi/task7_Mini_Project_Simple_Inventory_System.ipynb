# Task 7: Mini Project : Simple Inventory System (OOP)

```python
# Class Product
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.price=price
        self.category=category

    def __str__(self):
        return f"Product ({self.name}, {self.price}, {self.category})"

    def __add__(self,other):
        return self.price + other.price


# Inventory class
class Inventory:
    def __init__(self):
        self.products = []  # create a list to store products

    # Add product to inventory
    def add_product(self,product):
        self.products.append(product)
        print(product.name , "added to inventory") 

    # Remove product from inventory
    def remove_product(self,name):
        for product in self.products:
            if product.name==name:
                self.products.remove(product)
                print(name, "removed to inventory")  
                return
        print(name, "not found in inventory")  

    # Sum prices of all products
    def get_total_value(self):
        total = 0
        for product in self.products:
            total += product.price
        return total

    # Print info for each product
    def show_all_products(self):
        print("\nAll Products on Inventury")
        for product in self.products:
            print(product)


# Store class
class Store:    
    def __init__(self,store_name):
        self.store_name=store_name
        self.inventory=Inventory()  # Inventory object

    # Add new add product to inventory
    def add_new_product(self,name,price,category):
        product=Product(name,price,category)
        self.inventory.add_product(product)

    # show summary of store
    def show_summary(self):
        print("\nStore Name:", self.store_name)
        print("Total Items :", len(self.inventory.products))
        print("Total Inventory Value:",self.inventory.get_total_value())
        self.inventory.show_all_products()    

```

### Testing

```python
# Create object
store=Store("My Mart")
```

```python
# Adding 3 products
store.add_new_product("Laptop",50000,"Electronics")
store.add_new_product("Mouse",800,"Accessories")
store.add_new_product("Keyword",1000,"Accessories")
```

```python
# Showing summary
store.show_summary()
```
