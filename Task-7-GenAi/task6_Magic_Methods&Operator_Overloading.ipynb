# Task 6: Magic Method & Operator Overloading

```python
# Create a class
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.price=price
        self.category=category

    # Magic Method for readable string
    def __str__(self):
        return f"Product: ({self.name}, {self.price}, {self.category})"

    # Operating overloading for +
    def __add__(self,other):
        return self.price + other.price

```

```python
# Create object
laptop=Product("Laptop",50000,"Electronics")
mouse=Product("Mouse",700,"Accessories")
```

```python
# Testing for __str__()
print(laptop)
print(mouse)
```

```python
# Testing for __add__()
total_price = laptop + mouse
print("Total Price of Products:",total_price)
```
