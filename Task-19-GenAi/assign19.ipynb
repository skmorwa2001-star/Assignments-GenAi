## PART 1 - Introduction to word embeddings

### Task 1: Understanding Word Embeddings (Conceptual)

```python
'''
1. What are word embeddings ?
- Word embeddings are numerical representations of words in the form of dense vectors that capture their meanings and relationship.
- They help computers understand similarities between words in Natural Language Processing (NLP)

Example : 
       - king  --->  [0.8 , 0.3 , 0.4]
       - queen --->  [0.7 , 0.4 , 0.2]
       - apple  ---> [0.1 , 0.9 , 0.6]

- king , queen may have similar vector representations because their meanings are related       
'''
```

```python
'''
2. Why one-hot and bow fail to capture semantics ?
- One-Hot Encoding and Bag of Words (BoW) fail to capture semantics because they represent words using binary values
or word counts without understanding their meanings or relationships

Example : 
         - I like a car
         - I like a automobile

- Although car and automobile have similar meanings, one-hot encoding and BoW treat them as separate words and do not
automatically recognize their semantic similarity         
'''
```

```python
'''
3. How word embeddings solve these problems ?
- Word embeddings solve these problems by representing words as dense numerical vectors that capture semantic meaning and relationships between words
- Words with similar meanings have similar vector representations
'''
```

## PART 2 - Word2Vec Overview & Techniques 

### Task 2 : Word2Vec Overview

```python
'''
1. Explain what Word2Vec is ?
- Word2Vec is a popular NLP technique that converts words into dense numerical vectors called word embeddings.
- It learn semantic relationships between words by anaalyzing their surrounding context in a large text corpus

'''
```

```python
'''
2. Explain the idea of predicting words from context ?
- word prediction fron context is the process of predicting a word based on the surrounding words in a sentence
- Word2vec uses this idea to learn word embeddings and capture relationships between words

Example : sent ---> The cat drinks ___ .
                  -> the model may predict "milk" because it frequently appear in a similar context
'''
```

```python
'''
3. Define:
- Vocabulary : Vocabulary is the collection of all unique words present in a text corpus uses to train an NLP model

Example : I like AI and I like ML
Vocabulary : {I , like , AI , and , ML}

- Context Window : Context window is the number of surrounding words considered by a model to learn or predict a target word

Example : The cat drinks milk
If the target word is drinks and the window size is 1 the context words are cat and milk

- Embedding Dimensions : Embedding dimensions refers to the number of numerical values used to represent a word as a vector

Example : if the embedding dimension is 3:
   king --> [0.5 , 0.4 , 0.8]
   the word king is represent using 3 numerical values
'''
```

### Task 3 : Types of Word2Vec Techniques

```python
'''
1. CBOW (Continuous Bag of Words) : CBOW predicts a target word using its surrounding context words

Example : The cat drinks milk
Context words : the cat milk
target word : drinks
- CBOW is generally faster to train and works well with large datasets and frequent words
- it is good choice when training efficiency is important

2. Skip Gram : Skip gram predicts surrounding context words using a target word

Example : The cat drinks milk
target word : drinks
context word : cat , milk
- skip gram is often useful for learning representations of rare words 
- it can work well when you want good word representations and have enough training time and data
'''
```

### Task 4 : Neural Network Intuition Behind Word2Vec

```python
'''
1. Input layer representation : the input layer receives a word represented as a one-hot encoded vector
- the vector contains 1 at the positions of the input word and 0 at all other positions

2. Hidden layer (Embedding Layer) : the hidden layer converts the one hot vector into in similar , dense numerical vector called a word embedding
- its values are obtained from the network's weight matrix

3. Output layer : The output layer predicts a word based on the model's objective 
- CBOW : Predicts the target word from context words
- Skip gram : predicts context words from the target word

4. How do weights become word embeddings ?
- During training , the neural network adjusts its weights to improve word predictions. The weights connecting the input representation to the hidden layer from the embedding matrix
- Each words corresponding row in this matrix represents its learned embedding 
- Words appearing in similar contexts may develop similar vector representations 
'''
```

## PART 3 - Training Word2Vec on Custom Data

### Task 5 : Prepare Text for Word2Vec

```python
from nltk.tokenize import sent_tokenize , word_tokenize
```

```python
import pandas as pd

text=pd.read_csv('final_cleaned_data.csv')
```

```python
type(text)
```

```python
text=text['final_clean_text'].dropna().astype(str)
```

```python
# 1. tokenize cleaned text into sentences

sentences=[]

for i in text:
    sentence=sent_tokenize(i)
    sentences.append(sentence)
```

```python
sentences[:5]
```

```python
# 2. tokenize sentences into words

tokenized_sentences=[]

for i in text:
    sentence=word_tokenize(i)
    tokenized_sentences.append(sentence)
```

```python
# 3. Store them as a list into words
#  
print(tokenized_sentences[:3])
```

### Task 6: Train Word2Vec Model

```python
import gensim
```

```python
# 1. Train a Word2Vec model 

model=gensim.models.Word2Vec(vector_size=100,window=5,min_count=1,sg=0)
```

```python
model
```

```python
model.build_vocab(tokenized_sentences)
```

```python
model.train(tokenized_sentences,total_examples=model.corpus_count,epochs=model.epochs)
```

```python
# 2. Print

print("Vocabulary Size: ",len(model.wv))

print("Embedding vector for a sample word --> long :\n",model.wv['long'])
```

### Task 10 : Observations & Limitations

```python
'''
1. Difference between CBOW & SkipGram in practice
- CBOW (Continuous Bag of Words) predicts a target word from its surrounding context words, and Skip gram predicts surrounding context words from a target word
- CBOW is useful when large dataset and faster training
- Skip gram is used when rare words and detailed word representations
'''
```

```python
'''
2. Advantage of Word2Vec over TF-IDF
- Semantic similarity :- it can learn that word like car and automobile have related meainings
- Dense represntation :- it used compact vectors instead of typically sparse TF-IDF vectors
- Word relationship :- it can capture relationships between words based on their usage in context
- Generalization :- learned word vectors can be reused in different NLP tasks
- Context patterns :- it learns associations from surrounding words in a text corpus
'''
```

```python
'''
3. Limitations of Word2Vec

- Context independent embeddings
- Out of vocabulary words
- Requires large data
- Ignores word order
- rare words
- no true understanding
'''
```

```python
'''
4. Why context still matters in modern NLP (lead in to transformers)
- word meaning
- sentence understanding
- ambiguity
- better predictions

- unlike Word2Vec , which assigns a fixed vector to each word transformer models use mechanism such as self attention to create
context dependent representations 
- this allows the representation of a word to change according to the sentence in which it appears
'''
```
